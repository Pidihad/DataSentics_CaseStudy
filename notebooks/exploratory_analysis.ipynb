```python
import pandas as pd
from data_load import df, books, ratings, users

pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", 100)
```

```python
# DQ check

# Duplicity
print(f"Plné duplicity v df: {df.duplicated().sum()}")

# Duplicitní (ID, ISBN) - jeden uživatel hodnotil stejnou knihu vícekrát
dup_user_book = df.duplicated(subset=["User-ID", "ISBN"]).sum()
print(f"Duplicitní páry (User-ID, ISBN): {dup_user_book}")

# Null hodnoty
print("\nNull hodnoty:")
print(df.isnull().sum().sort_values(ascending=False))

# Null v procentech
print("\nNull v procentech:")
print((df.isnull().sum() / len(df) * 100).round(2).sort_values(ascending=False))
```
